# Data Augmentation

```python
import matplotlib.pyplot as plt
import numpy as np
import os
import random
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.preprocessing.image import ImageDataGenerator
%matplotlib inline
```

```python
def plotImages(images_arr):
    fig, axes = plt.subplots(1, 10, figsize=(20,20))
    axes = axes.flatten()
    for img, ax in zip(images_arr, axes):
        ax.imshow(img)
        ax.axis('off')
    plt.tight_layout()
    plt.show()
```

```python
gen = ImageDataGenerator(rotation_range=10, width_shift_range=0.1,
                         height_shift_range=0.1, shear_range=0.15, zoom_range=0.1,
                         channel_shift_range=10, horizontal_flip=True)
```

```python
chosen_image = random.choice(os.listdir(r'C:/Users/pc/Documents/A My Program/A Machine Learning/Kaggle Datasets/dogs-vs-cats/train/train/cat'))
```

```python
image_path = r'C:/Users/pc/Documents/A My Program/A Machine Learning/Kaggle Datasets/dogs-vs-cats/train/train/cat/' + chosen_image
```

```python
assert os.path.isfile(image_path)
```

```python
image = np.expand_dims(plt.imread(image_path), 0)
plt.imshow(image[0])
plt.show()
```

```python
aug_iter = gen.flow(image)
```

```python
aug_images = [next(aug_iter)[0].astype(np.uint8) for i in range(10)]
```

```python
plotImages(aug_images)
```
