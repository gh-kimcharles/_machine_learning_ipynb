```python
import numpy as np
your_word = "country"
ms = nlp.vocab.vectors.most_similar(
    np.asarray([nlp.vocab.vectors[nlp.vocab.strings[your_word]]]),
    n=10
)
words = [nlp.vocab.strings[w] for w in ms[0][0]]
distances = ms[2]
print(words)
```

```python
doc1 = nlp("This user likes to code and program applications.")
doc2 = nlp("The developer is creating a software using C++.")
print(doc1, "<->", doc2, doc1.similarity(doc2))
```

```python
doc3 = nlp("He is eating ice cream.")
print(doc1, "<->", doc3, doc1.similarity(doc3))
```
